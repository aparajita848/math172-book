__Mathematical Model:__ 

1. Let $N(t)$ and $C(t)$ represent the population of Nile perch, and cichlid fish at time $t$. Write a differential equation that models the population growth of the two fish species, in Lake Victoria, with the following __assumptions__:

- In the absence of harvesting or the predator Nile perch, the cichlids grow logistically with intrinsic growth rate rCand carrying capacity KC. Then, the cichlid population would grow exponentially until it reached the maximum density of the lake, or carrying capacity.
- We assume that the Nile perch is completely dependent on the cichlid as its only food source. As a consequence, in the absence of cichlids, the Nile perch population with decay with an intrinsic death rate dN.
- Additionally, add the harvesting rates to your equations with the assumption that the Nile perch and cichlids are each being harvested proportional to their population sizes.

2. Define each of the __variables__ and __parameters__ for your system:
- $N(t)$ - population of the Nile perch ($N$) at time $t$
- $C(t)$ - population of the cichlids ($C$) at time $t$
- $r$ - intrinsic growth rate of the cichlid population
- $d$ - intrinsic mortality rate of the Nile perch population in the absence of cichlids.
- $K$ - carrying capacity of the cichlid population
- $\alpha$ - capture efficiency: the effect of a single Nile perch on the per capita growth rate of the cichlid population.
- $\beta$ - conversion efficiency: the ability of Nile perch to convert each new cichlid into additional per capita growth rate for the Nile perch population.
- $h_N$ - harvesting rate of the Nile perch population 
- $h_C$ - harvesting rate of the cichlid population
  
3. Differential Equation System:

$$ \dfrac{dN}{dt}=-dN+\beta CN-Nh_N  $$

$$ \dfrac{dC}{dt} =r C\left(1-\dfrac{C}{K}\right)-\alpha CN-Ch_C $$

The first equation represents the population of Nile perch at time $t$, considering the logistic growth rate, variable harvesting, and conversion efficiency, including the intrinsic death rate assuming that the cichlid is the Nile perch’s only food source.
While, the second equation represents the population of cichlids at time $t$ in a logistic growth model, accounting for carrying capacity, intrinsic growth rate, capture efficiency, and harvesting rate.


```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
#from scipy.integrate import solve_ivp
import ipywidgets as widgets
from IPython.display import display, Math


dt = 1e-4
t = np.arange(0,200,dt)
N = np.linspace(0,200,100)
C = np.linspace(0,200,100)

#differential equation function
def Lake_Victoria_Sys(N,C,r,d,K,hc,hn,alpha,beta):
    return -d*N + beta*C*N -hn*N, r*C*(1-(C/K)) - alpha*C*N - hc*C
    

#function to calculate the population in time
def Fish_pop(t,N0,C0,r,d,K,hc,hn,alpha,beta):
    N_t = [N0] #initial population of Nile perch
    C_t = [C0] #initial population of cichlids
    for i in range(len(t)-1):
        N_t.append(N_t[i]+Lake_Victoria_Sys(N_t[i],C_t[i],r,d,K,hc,hn,alpha,beta)[0])
        C_t.append(C_t[i]+Lake_Victoria_Sys(N_t[i],C_t[i],r,d,K,hc,hn,alpha,beta)[1])
    
    return N_t,C_t

#setup plot
output = widgets.Output()
with output:
    fig, (ax1, ax2) = plt.subplots(1,2,constrained_layout=True, figsize=(8,4))
fig.canvas.toolbar_position = 'bottom'
fig.suptitle('Nile perch & Cichlids in Lake Victoria')

#Max_node, = ax1.plot([50],[25], 'co', label='Max growth')
# zero1, = ax1.plot([0],[0], 'bo', label='zero')
# zero2, = ax1.plot([100],[0], 'bo', label='zero')
line1_N, = ax1.plot(N, Lake_Victoria_Sys(N,C,0.1,0.07,100,0.05,0.05,0.001,0.0005)[0], 'c-', label='dN/dt')
line1_C, = ax1.plot(C, Lake_Victoria_Sys(N,C,0.1,0.07,100,0.05,0.05,0.001,0.0005)[1], 'm-', label='dC/dt')
ax1.grid(True)
ax1.set_ylim(0,5)
ax1.set_xlim(0,25)
ax1.set_xlabel('Cichlid (C)') #prey
ax1.set_ylabel('Nile perch (N)') #predator
ax1.set_title('Phase-line Diagram')
ax1.legend(loc='upper right')

line2_N, = ax2.plot(t,Fish_pop(t,50,50,0.1,0.1,100,0.05,0.05,0.001,0.0005)[0], 'c-', label='Nile perch')
line2_C, = ax2.plot(t,Fish_pop(t,50,50,0.1,0.1,100,0.05,0.05,0.001,0.0005)[1], 'm-', label='Cichlid')
ax2.set_ylim(0,50)
ax2.set_xlim(0,5)
ax2.grid(True)
ax2.set_xlabel('Time')
ax2.set_ylabel('Population Size')
ax2.set_title('Population Growth function')
ax2.legend(loc='upper right')

# create widget control elements
r_val = widgets.FloatSlider(value=0.1, min = 0.00, max = 1.00, step = 0.01, description = r'$r$')
d_val = widgets.FloatSlider(value=0.1, min = 0.00, max = 1.00, step = 0.01, description = r'$d$')
K_val = widgets.IntSlider(value=100, min = 0, max = 100, step = 1, description = r'$K$')
hc_val = widgets.FloatSlider(value=0.05, min = 0.00, max = 1.00, step = 0.01, description = r'$h_C$')
hn_val = widgets.FloatSlider(value=0.05, min = 0.00, max = 1.00, step = 0.01, description = r'$h_N$')
alpha_val = widgets.FloatSlider(value=0.001, min = 0.00, max = 1.00, step = 0.01, description = r'$\alpha$')
beta_val = widgets.FloatSlider(value=0.0005, min = 0.00, max = 1.00, step = 0.01, description = r'$\beta$')
N_0 = widgets.IntSlider(value = 50, min = 0, max = 200, step = 1, description = r'$N_0$')
C_0 = widgets.IntSlider(value = 50, min = 0, max = 200, step = 1, description = r'$C_0$')

def update(r,K,h,N0):    
    line1_N.set_ydata(Lake_Victoria_Sys(N,C,r,d,K,hc,hn,alpha,beta)[0])
    line1_C.set_ydata(Lake_Victoria_Sys(N,C,r,d,K,hc,hn,alpha,beta)[1])
    line2_N.set_ydata(Fish_pop(t,N0,C0,r,d,K,hc,hn,alpha,beta)[0])
    line2_N.set_ydata(Fish_pop(t,N0,C0,r,d,K,hc,hn,alpha,beta)[1])
    #Max_node.set_data([K/2],[r*K/4])
    #ax1.set_ylim(0, 1.5*r*K/4)
    #ax2.set_ylim(0,1.5*max(N_sol(t,N0,r,K,h)))
    fig.canvas.draw_idle()

# interactive_output matches dict keys to update()'s argument names
par_vals = widgets.interactive_output(update, {
    'r': r_val, 'd': d_val, 'K': K_val, 'hc': hc_val, 'hn': hn_val, 'alpha': alpha_val, 'beta': beta_val, 'N0': N_0, 'C0': C_0
})

disp_eqn = widgets.HTMLMath(
    value = r"""$$\begin{aligned}
    \dfrac{dN}{dt}=&-dN+\beta CN-h_NN\\
    \dfrac{dC}{dt}=&rc\left(1-\dfrac{C}{K}\right)-\alpha CN- h_CC
    \end{aligned}$$""",
    description= "Differential equation:"
)
disp_eqn.layout.margin = '0px 10px 10px 0px'
disp_eqn.layout.padding='5px 5px 5px 5px'
disp_eqn.layout.width = '350px'
disp_eqn.layout.border = 'solid 2.5px green'

par_vals = widgets.VBox([widgets.Label('Parameter values:'), r_val, d_val, K_val, hc_val, hn_val, alpha_val, beta_val, N_0, C_0])
par_vals.layout.margin = '0px 10px 10px 0px'
par_vals.layout.padding='5px 5px 5px 5px'
par_vals.layout.width = '350px'
par_vals.layout.border = 'solid 2px green'


widgets.HBox([par_vals,disp_eqn])
```
